```python
import math

# Prompt the user for input and convert the text into decimal numbers (floats)
x1 = float(input("Enter x1: "))
x2 = float(input("Enter y1: "))
y1 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

diff_x = x2 - x1
diff_y = y2 - y1

# This code is to
distance = math.sqrt(pow(diff_x, 2) + pow(diff_y, 2))

print("The distance between the two points is:", distance)
```
